```python
import numpy as np

np.random.seed(42)
n_transactions = 1_000_000

client_id = np.random.randint(1000, 5000, size=n_transactions).astype(np.float64)
product_id = np.random.randint(1, 100, size=n_transactions).astype(np.float64)
amount = np.random.uniform(1000, 50000, size=n_transactions)
quantity = np.random.randint(1, 10, size=n_transactions).astype(np.float64)
discount = np.random.uniform(0, 30, size=n_transactions)
region_code = np.random.randint(1, 5, size=n_transactions).astype(np.float64)

df = np.column_stack([client_id, product_id, amount, quantity, discount, region_code])

print("--- 1-КЕЗЕҢ ---")
print("Деректер массивінің өлшемі:", df.shape)

df[101] = df[100]
df[50, 2] = np.nan
df[500, 2] = np.nan
df[1000, 2] = 5_000_000.0

nan_count = np.isnan(df).sum()
print("\n--- 2-КЕЗЕҢ ---")
print("Жетіспейтін мәндер (NaN) саны:", nan_count)

valid_amounts = df[~np.isnan(df[:, 2]), 2]
median_val = np.median(valid_amounts)
df[np.isnan(df[:, 2]), 2] = median_val

df = np.unique(df, axis=0)

mean_art = np.mean(df[:, 2])
std_art = np.std(df[:, 2])
z_scores = (df[:, 2] - mean_art) / std_art
clean_df = df[np.abs(z_scores) <= 3]

high_value_sales = clean_df[clean_df[:, 2] > 20000]

print("\n--- 3-КЕЗЕҢ ---")
print("Тазаланған деректердің өлшемі:", clean_df.shape)
print("20,000-нан жоғары транзакциялар саны:", len(high_value_sales))

net_revenue = clean_df[:, 2] * (1 - clean_df[:, 4] / 100)
total_revenue = np.sum(net_revenue)

region_sales = {}
for r in range(1, 5):
    region_sales[r] = np.sum(net_revenue[clean_df[:, 5] == r])

print("\n--- 4-КЕЗЕҢ ---")
print(f"Жалпы таза табыс: {total_revenue:,.2f} тг")
for r, val in region_sales.items():
    print(f"  {r}-аймақтың сатылымы: {val:,.2f} тг")

amounts = clean_df[:, 2]
print("\n--- 5-КЕЗЕҢ: Сатып алу сомасының статистикасы ---")
print("Орташа мән (Mean):", np.mean(amounts))
print("Медиана (Median):", np.median(amounts))
print("Минимум (Min):", np.min(amounts))
print("Максимум (Max):", np.max(amounts))
print("Стандартты ауытқу (Std):", np.std(amounts))
print("Квартильдер (25%, 50%, 75%):", np.percentile(amounts, [25, 50, 75]))
```
